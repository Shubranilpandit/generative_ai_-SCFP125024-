```python
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
from langchain_groq import ChatGroq
```

```python
import os
```

```python
model = "openai/gpt-oss-20b"
api_key = os.environ["GROQ_API_KEY"]
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.environ["GROQ_API_KEY"])
```

```python
llm.invoke("hii")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
input_temp=ChatPromptTemplate.from_messages([
    ("system","you are a exprience  poet.write meaningful poem.do not response query other than poem"),
    ("user","write a poem on {poem_topic}"),
])
```

```python
input=input_temp.invoke({"poem_topic":"rain fall"})
```

```python
response=llm.invoke(input)
```

```python
response.content
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
parser.invoke(response)
```

```python
input=input_temp.invoke({"poem_topic":"rain fall"})
response=llm.invoke(input)
final_output=parser.invoke(response)
print(final_output)
```

```python
chain=input_temp | llm | parser
```

```python
chain.invoke({"poem_topic":"rain fall"})
```

```python
from langchain_core.output_parsers import JsonOutputParser
```

```python
json_parser = JsonOutputParser()
```

```python
json_parser.get_format_instructions()
```

```python
summarize_temp = ChatPromptTemplate.from_messages([
    ("system", "You are expert in summarizing poem. Summarize poem in 5 points {JSON_FORMATING}"),
    ("user", "Summarize the poem {poem}"),
]).partial(JSON_FORMATING=json_parser.get_format_instructions())
```

```python
summarize_temp
```

```python
final_response=chain.invoke({"poem_topic":"rain fall"})
```

```python
summarize_temp.invoke({"poem":final_response})
```

```python
step_2_input = summarize_temp.invoke({"poem":final_response})
```

```python
step_2_resp = llm.invoke
```

```python
step_1_input = input_temp.invoke({"poem_topic":"rain fall"})

step_1_resp = llm.invoke(step_1_input)

final_response_1=chain.invoke(step_1_resp)

step_2_input = summarize_temp.invoke({"poem":final_response})

step_2_resp = llm.invoke(step_2_input)

final_response_2=chain.invoke(step_2_resp)
```

```python
chain = input_temp | llm | parser | summarize_temp | llm | parser
```

```python
chain.invoke({"poem_topic":"rain fall"})
```
