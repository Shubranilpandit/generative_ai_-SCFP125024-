```python
from groq import Groq
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
import os
```

```python
client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

```python
messages=[
        {
            "role": "user",
            "content": "Explain the importance of fast language models",
        }
    ]
```

```python
messages
```

```python
chat_completion = client.chat.completions.create(messages=messages, model="openai/gpt-oss-20b")
```

```python
print(chat_completion.choices[0].message.content)
```

```python
def ask_llm(user_query):
    message=[
    {
        "role":"user"
        ,"content":user_query
        
    }
]
    chat_completion = client.chat.completions.create(messages=message, model="openai/gpt-oss-20b")
    response=chat_completion.choices[0].message.content
    return response
```

```python
ask_llm("Explain the importance of fast language models")
```

```python
history = []
while True:
    user_query = input("User: ")
    if user_query.lower() == "exit":
        break
    message = {
        "role": "user",
        "content": user_query
    }
    history.append(message)
    chat_completion = client.chat.completions.create(messages=history, model="openai/gpt-oss-20b")
    response = chat_completion.choices[0].message.content
    print("AI: " + response)
    ai_response = {
        "role": "assistant",
        "content": response
    }
    history.append(ai_response)
```
